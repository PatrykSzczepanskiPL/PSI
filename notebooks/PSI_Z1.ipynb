```python
import numpy as np

```

```python
#przyklad 1
x = np.array([[0.5],[0.75],[0.1]])
W = np.array([
     [0.1,0.1,-0.3],
     [0.1,0.2,0.0],
     [0.0,0.7,0.1],
     [0.2,0.4,0.0],
     [-0.3,0.5,0.1]])

print(x)
print(W)
print(np.dot(W,x))
```

```python
#pzyklad 2
Wh = W.copy()
Wy = np.array([
    [0.7,0.9,-0.4,0.8,0.1],
    [0.8,0.5,0.3,0.1,0.0],
    [-0.3,0.9,0.3,0.1,-0.2]
])
print(Wh)
print(Wy)
layer1_hidden = np.dot(Wh,x)
layer_output = np.dot(Wy,layer1_hidden)
print(layer1_hidden)
print(layer_output)
```

```python
#ZADANIE 1
def neuron(inputs: np.ndarray,weights : np.ndarray, bias : float) -> float:
    try:
        output = np.sum(np.dot(inputs,weights))
        output += bias
    except:
        output = 0
        print("couldn't calculate output")
    return output
out = neuron(W,x,0)
print(out)
```

```python
#zadanie 2
def neural_network(inputs: np.ndarray, weights: np.ndarray) -> np.ndarray:
    if inputs.shape[0] != weights.shape[0]:
        print("inputs row number and weights column number must be equal")
        return
    try:
        out = np.dot(inputs,weights)
    except:
        print("couldn't calculate output")
        return
    return out

print(neural_network(x,W))

```

```python
#   ZADANIE 3

def deep_neural_network(inputs: np.ndarray, weights_list: list[np.ndarray]) -> np.ndarray:
    if inputs.shape[0] != weights_list[0].shape[1]:
        print("inputs row number and weights column number must be equal")
        return
    out = np.dot(weights_list[0],inputs)
    for i in range(len(weights_list) - 1):
        if out.shape[0] != weights_list[i + 1].shape[1]:
            print("inputs row number and weights column number must be equa12321l")
            return
        out = np.dot(weights_list[i+1],out)
    return out

print(deep_neural_network(x,[Wh,Wy]))
```

```python
class Dense_Neural_Network:

    def __init__(self, input_dim:int):
        self.weights_list:list[np.ndarray] = list()
        self.input_dim:int = input_dim
        pass

    def add_layer(self, n, weight_min_value=None, weight_max_value = None, weights :np.ndarray = None):
        if weights is None:
            weights = np.random.uniform(weight_min_value, weight_max_value, (n, self.input_dim))
        if(weights.shape[0] != n):
            print("weights row number and weights column number must be equal")
            return
        self.weights_list.append(weights)
        self.input_dim = n

    def predict(self, inputs) -> np.ndarray:
        if inputs.shape[0] != self.weights_list[0].shape[1]:
            print("inputs row number and weights column number must be equal")
            return
        out = np.dot(self.weights_list[0], inputs)
        for i in range(len(self.weights_list) - 1):
            if out.shape[0] != self.weights_list[i + 1].shape[1]:
                print("inputs row number and weights column number must be equal")
                return
            out = np.dot(self.weights_list[i + 1], out)
        return out

    def load_weights(self, file_name):
        fp:np.ndarray = np.load(file_name)
        self.add_layer(fp.shape[0],None,None,fp)


model = Dense_Neural_Network(3)
model.add_layer(5,None,None,Wh)
model.add_layer(3,None,None,Wy)
print(model.predict(x))
```
